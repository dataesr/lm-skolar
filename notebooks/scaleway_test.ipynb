```python
inference_api_url = "https://api.scaleway.com/inference/v1/regions/fr-par"

api_headers = {
    "Content-Type": "application/json",
    "X-Auth-Token": os.getenv("SCW_SECRET_KEY"),
}
# list models
response = requests.get(f"{inference_api_url}/models", headers=api_headers)

# create a model deployment
payload = {
    "project_id": "05e2110e-746a-4578-bccd-9ef7e93cfc0c",
    "name": "ack-inference",
    "model_id": "398a435e-c889-45fa-ba56-a099d9990f1a",
    "node_type_name": "L4",
    "endpoints": [{"public_network": {}}],
}
response = requests.post(
    f"{inference_api_url}/deployments",
    headers=api_headers,
    data=json.dumps(payload),
)

inference_url = response.json().get("endpoints", [{}])[0].get("url")
deployment_id = response.json().get("id")
deployment_status = response.json().get("status") 

# list deployments
response = requests.get(f"{inference_api_url}/deployments", headers=api_headers)

# get a deployment
response = requests.get(f"{inference_api_url}/deployments/{deployment_id}", headers=api_headers)

# delete a deployment 
response = requests.delete(f"{inference_api_url}/deployments/{deployment_id}", headers=api_headers)
```
